```python
import os
from pathlib import Path
import subprocess
import sys

url = "https://github.com/rewanebrahembadr-cmyk/GlazeVision"
root = Path("/content/GlazeVision")
if not root.exists():
    subprocess.run(["git", "clone", url, str(root)], check=True)
os.chdir(root)
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], check=True)
```

```python
metadata_only = False
command = [sys.executable, "scripts/download_data.py"]
if metadata_only:
    command.append("--metadata-only")
subprocess.run(command, check=True)
subprocess.run([sys.executable, "scripts/make_split.py"], check=True)
```
