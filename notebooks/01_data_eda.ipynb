```python
import matplotlib.pyplot as plt
from glazevision.data import joined_samples

frame = joined_samples()
display(frame.groupby(["split", "surface"], dropna=False).size().unstack(fill_value=0))
for column in ("surface", "image_surface", "atmosphere", "cone"):
    frame[column].fillna("Missing").value_counts().head(15).plot.bar(title=column)
    plt.tight_layout()
    plt.show()
assert frame.groupby("group_id").split.nunique().max() == 1
```
